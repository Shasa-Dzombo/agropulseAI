# Exploratory Data Analysis (EDA) for Crop Recommendation

This notebook performs an initial analysis of the synthetic dataset to understand its properties, distributions, and relationships between features.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Adjust path to load data from the correct location
DATA_PATH = os.path.join('..', 'data', 'raw_data.csv')

df = pd.read_csv(DATA_PATH)

print("Dataset Shape:", df.shape)
df.head()
```

## 1. Basic Information and Summary Statistics

```python
df.info()
```

```python
df.describe()
```

## 2. Missing Value Analysis

```python
missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df)) * 100
pd.DataFrame({'missing_values': missing_values, 'percentage': missing_percentage})
```

## 3. Target Variable Distribution

```python
plt.figure(figsize=(12, 6))
sns.countplot(y='crop', data=df, order = df['crop'].value_counts().index)
plt.title('Distribution of Crops in the Dataset')
plt.xlabel('Count')
plt.ylabel('Crop')
plt.show()
```

## 4. Numerical Feature Distributions

```python
numerical_cols = df.select_dtypes(include=['float64', 'int64']).columns

df[numerical_cols].hist(bins=30, figsize=(20, 15), layout=(3, 3))
plt.suptitle('Distribution of Numerical Features')
plt.show()
```

## 5. Correlation Analysis

```python
plt.figure(figsize=(12, 10))
sns.heatmap(df[numerical_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

## 6. Feature Relationships with Target Variable

```python
for col in numerical_cols:
    plt.figure(figsize=(15, 7))
    sns.boxplot(x='crop', y=col, data=df)
    plt.title(f'{col.capitalize()} by Crop')
    plt.xticks(rotation=45)
    plt.show()
```
